```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
equip_no_rep = pd.read_csv("../DATA/equip_no_rep.csv")
```

```python
columns=['possessionPct', 'foulsCommitted', 'accuratePasses', 'totalPasses', 'passPct',
    'totalLongBalls', 'accurateLongBalls', 'longballPct', 'effectiveTackles',
    'tacklePct', 'interceptions', 'effectiveClearance', 'totalClearance']
```

```python
for col in columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(data=equip_no_rep, x=col, kde=True, bins=30)
    plt.title(f'Distribution of {col}', fontsize=14, fontweight='bold')
    plt.xlabel(col, fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.tight_layout()
    plt.show() 
```
